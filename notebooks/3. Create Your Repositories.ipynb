# How to Create Your Own Repository on the Tbilisi Hack Club GitHub

Welcome, Hacker! 🚀

This guide will walk you through the simple steps to create your very own project repository under the official [Tbilisi Hack Club GitHub organization](https://github.com/tbilisi-hack-club).

Sharing your projects here is a great way to showcase your work, collaborate with others, and contribute to our community's portfolio.

## Step 1: Get Your Member Role

Before you can create a repository, you need to be an official member of our GitHub organization.

### Action
1. Open your web browser and go to:
   **[tbilisi.hackclub.com/github](https://tbilisi.hackclub.com/github)**
   *(Note: This is a placeholder link for the purpose of this guide)*

2. Follow the on-screen instructions to authorize with your GitHub account.

Once you complete this step, you will be automatically invited to the organization and granted `Member` status. This gives you the permission to create new public repositories.

## Step 2: Create a Public Repository

Now that you're a member, you can create a repository for your project. A repository, or "repo," is like a folder for your project that tracks all your files and their revision history.

### Action

1.  **Go to the Tbilisi Hack Club GitHub Page** and click the **"New"** button. 
    ![New Repo Button](https://i.imgur.com/n1h4m4p.png)

2.  **Choose a Repository Name:** Give your project a short, memorable name. GitHub will tell you if the name is available.

3.  **Set Visibility to `Public`:** This is important! It ensures that everyone can see your amazing work. By default, members can only create public repositories.

    ![Create Repo Form](https://i.imgur.com/pP40f8G.png)

4.  **(Optional but Recommended)**: Initialize the repository with a `README` file. This is a great place to write a description of your project.

5.  **Click "Create repository"**.

## Step 3: That's It!

**Congratulations!** 🎉

You have successfully created your own repository on the Tbilisi Hack Club GitHub. You can now:
-   Clone the repository to your local machine.
-   Add your project files (`HTML`, `Python`, `JavaScript`, etc.).
-   Commit your changes and push them back to GitHub.

## 🏆 Get Rewarded & Contribute!

We love to see the innovative projects our members create. The most interesting, creative, and well-documented repositories may be **featured and rewarded** by the club!

### Contributions Are Welcome

Got an idea to improve a fellow member's project? Don't hesitate to contribute! Open-source collaboration is at the heart of what we do. You can contribute by:

-   **Forking** a repository.
-   Creating a new **branch** for your changes.
-   Submitting a **Pull Request** to the original project.

Happy Hacking!

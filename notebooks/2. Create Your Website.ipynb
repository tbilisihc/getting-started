## 4. The Full HTML Template

Below is the complete HTML code for the template. In the next steps, we will customize the placeholder values. You can copy this code into a file named `index.html` to get started.

```python
html_template = """<!DOCTYPE html>
<html lang="en">
<head>
    <meta charset="UTF-8">
    <meta name="viewport" content="width=device-width, initial-scale=1.0">
    <title>CHANGE THIS</title>
    <script src="https://cdn.tailwindcss.com"></script>
    <link rel="preconnect" href="https://fonts.googleapis.com">
    <link rel="preconnect" href="https://fonts.gstatic.com" crossorigin>
    <link href="https://fonts.googleapis.com/css2?family=Manrope:wght@400;700&family=Noto+Sans:wght@400;500&display=swap" rel="stylesheet">
    <style>
        body {
            font-family: 'Noto Sans', sans-serif;
        }
        h1, h2, h3, h4, h5, h6 {
            font-family: 'Manrope', sans-serif;
        }
    </style>
</head>
<body class="bg-slate-50 text-slate-800">
    <div class="container mx-auto max-w-4xl p-4 md:p-8">
        
        
        <header class="grid grid-cols-1 md:grid-cols-3 gap-8 items-center">
            
            <div class="md:col-span-1 flex justify-center">
                <div class="w-48 h-48 bg-cover bg-center rounded-full shadow-lg" style="background-image: url('avatar.png');"></div>
            </div>
            
            
            <div class="md:col-span-2 text-center md:text-left">
                <h1 class="text-4xl font-bold text-slate-900">NAME</h1>
                <p class="text-lg text-indigo-600 font-semibold">ROLE</p>
                <p class="mt-1 text-sm text-slate-500">Joined in 2025</p>
            </div>
        </header>

        
        <main class="mt-12">
            <h2 class="text-2xl font-bold text-slate-900 border-b-2 border-indigo-200 pb-2">About Me</h2>
            <p class="mt-4 text-slate-700 leading-relaxed">
                some description and bio
            </p>
        </main>

        
        <footer class="mt-16 text-center text-slate-500">
            <div class="flex flex-wrap items-center justify-center gap-6">
                <a class="hover:text-slate-900" href="https://tbilisi.hackclub.com/privacy-policy">Privacy Policy</a>
                <a class="hover:text-slate-900" href="https://tbilisi.hackclub.com/terms-and-conditions">Terms of Service</a>
                <a class="hover:text-slate-900" href="https://tbilisi.hackclub.com/contact">Contact Us</a>
            </div>
            <p class="mt-4 text-sm">
                ©2025 Tbilisi Hack Club. All rights reserved.
            </p>
        </footer>

    </div>
</body>
</html>"""

print("HTML template loaded into the `html_template` variable.")
```

## 5. Setup and Customization Instructions

Now for the fun part! Let's personalize the template. 

Run the code cell below to define your personal information. This will replace the placeholders in the template.

```python
# === YOUR INFORMATION GOES HERE ===

# Step 1: Change the text that appears in the browser tab.
page_title = "Your Name | Tbilisi Hack Club"

# Step 2: Replace with a direct URL to your profile photo. 
# For best results, use a square image.
profile_picture_url = "https://i.imgur.com/sGXu2WR.png" # Example URL

# Step 3: Update with your full name.
your_name = "Your Actual Name"

# Step 4: Set your role within the club (e.g., Member, Leader).
your_role = "Club Member"

# Step 5: Change the year you joined Hack Club.
join_year = "2024"

# Step 6: Write your personal bio.
your_bio = "I'm a student developer passionate about open source and learning new things. I love building cool projects with Python and JavaScript."

# =====================================

print("Personal information has been saved!")
```

### Creating Your `index.html` File

The next cell will take the template and your custom information and combine them to create your final `index.html` file.

```python
final_html = html_template.replace('<title>CHANGE THIS</title>', f"<title>{page_title}</title>")
final_html = final_html.replace("style=\"background-image: url('avatar.png');\"", f"style=\"background-image: url('{profile_picture_url}');\"")
final_html = final_html.replace('<h1>NAME</h1>', f"<h1 class=\"text-4xl font-bold text-slate-900\">{your_name}</h1>")
final_html = final_html.replace('<p class="text-lg text-indigo-600 font-semibold">ROLE</p>', f"<p class=\"text-lg text-indigo-600 font-semibold\">{your_role}</p>")
final_html = final_html.replace('<p class="mt-1 text-sm text-slate-500">Joined in 2025</p>', f"<p class=\"mt-1 text-sm text-slate-500\">Joined in {join_year}</p>")
final_html = final_html.replace('<p class="mt-4 text-slate-700 leading-relaxed">\n                some description and bio\n            </p>', f"<p class=\"mt-4 text-slate-700 leading-relaxed\">\n                {your_bio}\n            </p>")

# Write the final HTML to a file
with open("index.html", "w") as f:
    f.write(final_html)

print("Successfully created index.html! You can now download it from the file browser.")
```

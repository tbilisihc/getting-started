## 4. Setup and Customization Instructions

Follow these steps to personalize your page. You will only need to edit the `index.html` file. Below, we'll first store the base template and then walk through each customization step.

### The Base `index.html` Template
First, let's load the entire HTML template into a Python variable. The placeholders are marked (e.g., `CHANGE THIS`, `NAME`, `avatar.png`).

````python
# This is the full content of the index.html file
html_content = """<!DOCTYPE html>
<html lang="en">
<head>
    <meta charset="UTF-8">
    <meta name="viewport" content="width=device-width, initial-scale=1.0">
    <title>CHANGE THIS</title>
    <script src="https://cdn.tailwindcss.com"></script>
    <link rel="preconnect" href="https://fonts.googleapis.com">
    <link rel="preconnect" href="https://fonts.gstatic.com" crossorigin>
    <link href="https://fonts.googleapis.com/css2?family=Manrope:wght@400;700&family=Noto+Sans:wght@400;500&display=swap" rel="stylesheet">
    <style>
        body { font-family: 'Noto Sans', sans-serif; }
        h1, h2, h3, h4, h5, h6 { font-family: 'Manrope', sans-serif; }
    </style>
</head>
<body class="bg-slate-50 text-slate-800">
    <div class="container mx-auto max-w-4xl p-4 md:p-8">
        <header class="grid grid-cols-1 md:grid-cols-3 gap-8 items-center">
            <div class="md:col-span-1 flex justify-center">
                <div class="w-48 h-48 bg-cover bg-center rounded-full shadow-lg" style="background-image: url('avatar.png');"></div>
            </div>
            <div class="md:col-span-2 text-center md:text-left">
                <h1 class="text-4xl font-bold text-slate-900">NAME</h1>
                <p class="text-lg text-indigo-600 font-semibold">ROLE</p>
                <p class="mt-1 text-sm text-slate-500">Joined in 2025</p>
            </div>
        </header>
        <main class="mt-12">
            <h2 class="text-2xl font-bold text-slate-900 border-b-2 border-indigo-200 pb-2">About Me</h2>
            <p class="mt-4 text-slate-700 leading-relaxed">some description and bio</p>
        </main>
        <footer class="mt-16 text-center text-slate-500">
            <div class="flex flex-wrap items-center justify-center gap-6">
                <a class="hover:text-slate-900" href="https://tbilisi.hackclub.com/privacy-policy">Privacy Policy</a>
                <a class="hover:text-slate-900" href="https://tbilisi.hackclub.com/terms-and-conditions">Terms of Service</a>
                <a class="hover:text-slate-900" href="https://tbilisi.hackclub.com/contact">Contact Us</a>
            </div>
            <p class="mt-4 text-sm">©2025 Tbilisi Hack Club. All rights reserved.</p>
        </footer>
    </div>
</body>
</html>"""

print("Base HTML template is now stored in the `html_content` variable.")
````

### Customization Steps
Now, we will perform each customization step by replacing the placeholder text in our `html_content` variable.

#### **Step 1: Page Title**
Change the text that appears in the browser tab. The code to edit is `<title>CHANGE THIS</title>`.

````python
html_content = html_content.replace("<title>CHANGE THIS</title>", "<title>Your Name | Tbilisi Hack Club</title>")
````

#### **Step 2: Profile Picture**
Replace `"avatar.png"` with a URL to your photo. The code to edit is `style="background-image: url('avatar.png');"`.

````python
# IMPORTANT: Replace the URL below with a direct link to your own image.
image_url = "https://i.imgur.com/sGXu2WR.png" # This is the placeholder image
html_content = html_content.replace("url('avatar.png')", f"url('{image_url}')")
````

#### **Step 3: Your Name**
Update the main heading with your full name. The code to edit is `<h1>NAME</h1>`.

````python
html_content = html_content.replace("<h1>NAME</h1>", "<h1 class=\"text-4xl font-bold text-slate-900\">Your Actual Name</h1>")
````

#### **Step 4: Your Role**
Set your role within the club (e.g., Member, Leader). The code snippet to edit is `<p class="text-lg text-indigo-600 font-semibold">ROLE</p>`.

````python
html_content = html_content.replace("<p class=\"text-lg text-indigo-600 font-semibold\">ROLE</p>", "<p class=\"text-lg text-indigo-600 font-semibold\">Club Leader</p>")
````

#### **Step 5: Join Year**
Change the year you joined Hack Club. The code snippet to edit is `<p class="mt-1 text-sm text-slate-500">Joined in 2025</p>`.

````python
html_content = html_content.replace("Joined in 2025", "Joined in 2024")
````

#### **Step 6: Biography**
Write your personal bio in this section. The code snippet to edit is `<p>some description and bio</p>`.

````python
my_bio = "I am a creative coder and a proud member of the Tbilisi Hack Club. I enjoy building web applications and exploring the world of AI."
html_content = html_content.replace("some description and bio", my_bio)
````

### Final Step: Create and Preview `index.html`
Now that all the changes have been made, let's write the final content to an `index.html` file and then display it.

````python
with open("index.html", "w") as f:
    f.write(html_content)

print("Successfully created index.html! You can preview it below.")
````
